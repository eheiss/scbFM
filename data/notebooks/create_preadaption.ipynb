```python
import anndata as ad
import numpy as np
```

```python
SEED = 1234
PRETRAIN_SIZE = 700000
```

```python
pretrain_bulk = ad.read_h5ad("/cluster/work/boeva/eheiss/datasets/preprocessed_bulk/pretraining_binned.h5ad")
```

```python
rng = np.random.default_rng(SEED)

idx = np.arange(pretrain_bulk.n_obs)
rng.shuffle(idx)

idx1 = idx[:PRETRAIN_SIZE]
idx2 = idx[PRETRAIN_SIZE:]

adata_pretrain = pretrain_bulk[idx1].copy()
adata_preadapt = pretrain_bulk[idx2].copy()
```

```python
adata_pretrain.write(f"/cluster/work/boeva/eheiss/datasets/preprocessed_bulk/pretrain_binned_{PRETRAIN_SIZE}.h5ad")
adata_preadapt.write("/cluster/work/boeva/eheiss/datasets/preprocessed_bulk/preadapt_binned.h5ad")
```
