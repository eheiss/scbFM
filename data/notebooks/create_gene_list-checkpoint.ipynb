```python
import anndata as ad
import pandas as pd
import numpy as np
import tables
import h5py
from scipy import sparse
from tqdm import tqdm
```

```python
with open("/cluster/work/boeva/eheiss/scbFM/data/gene_list.txt") as f:
    gene_list = [line.strip() for line in f if line.strip()]
gene_set = set(gene_list)
```

```python
tcga = ad.read_h5ad("/cluster/work/boeva/eheiss/datasets/TCGA/tcga.h5ad")
print(tcga)
tcga_genes = set(tcga.var_names)
```

```python
in_list_mask = np.array([g in gene_set for g in tcga.var_names])
not_in_list_mask = ~in_list_mask

X = tcga.X.tocsr() if sparse.issparse(tcga.X) else np.asarray(tcga.X)

if sparse.issparse(X):
    total_nonzero = X.getnnz(axis=1)
    not_in_list_nonzero = X[:, not_in_list_mask].getnnz(axis=1)

    total_reads = np.asarray(X.sum(axis=1)).ravel()
    not_in_list_reads = np.asarray(X[:, not_in_list_mask].sum(axis=1)).ravel()
else:
    total_nonzero = (X > 0).sum(axis=1)
    not_in_list_nonzero = (X[:, not_in_list_mask] > 0).sum(axis=1)

    total_reads = X.sum(axis=1)
    not_in_list_reads = X[:, not_in_list_mask].sum(axis=1)

frac_nonzero_not_in_list = np.divide(
    not_in_list_nonzero,
    total_nonzero,
    out=np.zeros_like(total_nonzero, dtype=float),
    where=total_nonzero > 0,
)

frac_reads_not_in_list = np.divide(
    not_in_list_reads,
    total_reads,
    out=np.zeros_like(total_reads, dtype=float),
    where=total_reads > 0,
)

print("Average portion of non-zero genes NOT in gene_list:",
      frac_nonzero_not_in_list.mean())

print("Average portion of total reads NOT in gene_list:",
      frac_reads_not_in_list.mean())
```

```python
path = "/cluster/work/boeva/eheiss/datasets/ARCHS4/human_gene_v2.latest.h5"

with h5py.File(path, "r") as f:
    print("expression:", f["data/expression"].shape, f["data/expression"].dtype)
    print("genes:", f["meta/genes/symbol"].shape, f["meta/genes/symbol"].dtype)
    print("samples:", f["meta/samples/sample"].shape, f["meta/samples/sample"].dtype)
    
with h5py.File(path, "r") as f:
    ensembl = f["meta/genes/ensembl_gene"][:]

# decode bytes → string
archs4_genes = set([e.decode() if isinstance(e, bytes) else e for e in ensembl])
```

```python
with h5py.File("/cluster/work/boeva/eheiss/datasets/ARCHS4/human_gene_v2.latest.h5", "r") as f:
    ensembl = f["meta/genes/ensembl_gene"][:]
    ensembl = np.array([e.decode() if isinstance(e, bytes) else e for e in ensembl])

    # optional: strip version suffix like ENSG000001234.5 -> ENSG000001234
    ensembl_base = np.array([g.split(".")[0] for g in ensembl])

    in_list_mask = np.array([g in gene_set for g in ensembl_base])
    not_in_list_mask = ~in_list_mask

    X = f["data/expression"]   # shape: genes x samples

    n_samples = X.shape[1]
    chunk_size = 2000

    frac_nonzero_list = []
    frac_reads_list = []

    for start in tqdm(range(0, n_samples, chunk_size)):
        end = min(start + chunk_size, n_samples)

        block = X[:, start:end]   # shape: genes x chunk_samples
        block = np.asarray(block)

        total_nonzero = (block > 0).sum(axis=0)
        not_in_list_nonzero = (block[not_in_list_mask, :] > 0).sum(axis=0)

        total_reads = block.sum(axis=0)
        not_in_list_reads = block[not_in_list_mask, :].sum(axis=0)

        frac_nonzero = np.divide(
            not_in_list_nonzero,
            total_nonzero,
            out=np.zeros_like(total_nonzero, dtype=float),
            where=total_nonzero > 0,
        )

        frac_reads = np.divide(
            not_in_list_reads,
            total_reads,
            out=np.zeros_like(total_reads, dtype=float),
            where=total_reads > 0,
        )

        frac_nonzero_list.append(frac_nonzero)
        frac_reads_list.append(frac_reads)

    frac_nonzero_all = np.concatenate(frac_nonzero_list)
    frac_reads_all = np.concatenate(frac_reads_list)

print("Average portion of non-zero genes NOT in gene_list:",
      frac_nonzero_all.mean())

print("Average portion of total reads NOT in gene_list:",
      frac_reads_all.mean())
```
